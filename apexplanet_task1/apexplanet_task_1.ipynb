```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/content/drive/MyDrive/ApexPlanet_DataAnalytics_Dataset.xlsx")
df.head()
```

```python
print(df.shape)

print(df.columns)

print(df.info())

print(df.describe(include='all'))
```

```python
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)
df = df.drop_duplicates()

print(df.shape)
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
print(df.dtypes)
```

```python
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["Gender"] = df["Gender"].str.title()

df["City"] = df["City"].str.title()

df["Category"] = df["Category"].str.title()

df["Product"] = df["Product"].str.title()
```

```python
df["Age"].fillna(df["Age"].median(), inplace=True)

df["Gender"].fillna(df["Gender"].mode()[0], inplace=True)

df["City"].fillna(df["City"].mode()[0], inplace=True)

df["Quantity"].fillna(df["Quantity"].median(), inplace=True)

df["Unit_Price"].fillna(df["Unit_Price"].median(), inplace=True)

df["Total_Sales"].fillna(df["Total_Sales"].median(), inplace=True)
```

```python
df["Month"] = df["Order_Date"].dt.month_name()
df["Year"] = df["Order_Date"].dt.year
df["Day"] = df["Order_Date"].dt.day_name()
df["Calculated_Sales"] = df["Quantity"] * df["Unit_Price"]

df["Difference"] = df["Calculated_Sales"] - df["Total_Sales"]

print(df["Difference"].describe())
df["Total_Sales"] = df["Calculated_Sales"]
df.drop(["Calculated_Sales","Difference"],axis=1,inplace=True)
```

```python
Q1 = df["Total_Sales"].quantile(0.25)

Q3 = df["Total_Sales"].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR

upper = Q3+1.5*IQR

outliers = df[(df["Total_Sales"]<lower) | (df["Total_Sales"]>upper)]

print("Outliers :",len(outliers))
```

```python
print(df.info())

print(df.head())

print(df.describe())
```

```python
df.to_excel("Cleaned_Sales_Dataset.xlsx",index=False)

print("Dataset Saved Successfully")
```
